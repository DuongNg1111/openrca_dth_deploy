Loads the built-in **mock** telemetry and plots one KPI. In M3, swap in the real OpenRCA dataset.


```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
from src.input_module.telemetry_loader import load_mock
tele = load_mock()
print('rows:', len(tele['metric']))
tele['metric'][:3]
```

```python
# Plot the signature KPI for each component (needs matplotlib: pip install matplotlib)
import matplotlib.pyplot as plt
KPI = 'disk_io_read'
by_comp = {}
for r in tele['metric']:
    if r['kpi'] == KPI:
        by_comp.setdefault(r['component'], []).append(r['value'])
plt.figure(figsize=(8,4))
for comp, ys in by_comp.items():
    plt.plot(ys, marker='o', label=comp)
plt.title(f'{KPI} over time'); plt.legend(); plt.xlabel('minute'); plt.ylabel(KPI)
plt.show()
```
